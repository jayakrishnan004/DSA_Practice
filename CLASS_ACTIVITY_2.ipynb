A hospital wants to develop a simple appointment management system to manage doctor
appointments for patients.
Create a Python class named Doctor with the following attributes:
* doctor_id
* doctor_name
* specialization
* available_slots
Create another class named AppointmentSystem to manage doctors and patient appointments.
Requirements :
1. Book Appointment

The system should allow a patient to book an appointment with a doctor.
An appointment can be booked only when:
* The doctor exists.
* The requested slot is available.
* The patient has not already booked an appointment with the same doctor.

When successfully booked:
* display - Appointment Booked
* If the doctor does not exist - Doctor Not Found
* If the slot is unavailable - Slot Not Available
* If the patient has already booked the same doctor - Appointment Already Exists
2. Cancel Appointment

A patient should be able to cancel an existing appointment
* If the appointment exists.
* Remove the appointment.
* Make the slot available again.
* Display - Appointment Cancelled
* If the appointment does not exist - Appointment Not Found

3. Display Doctor Details

Display:
* Doctor ID
* Doctor Name
* Specialization
* Number of available slots

4. Display Patient Appointments
* Display all appointments booked by a particular patient.
* If the patient has no appointments - No Appointments

5. Appointment Summary

Display:
* Total number of appointments
* Doctor with the highest number of appointments
* Number of available slots across all doctors

Input Format

* The first line contains the number of doctors N.

The next N lines contain:
* Doctor_ID Doctor_Name Specialization Number_of_Slots
* The next line contains the number of operations M.

Each operation will be one of:
* BOOK Patient_Name Doctor_ID Slot
* CANCEL Patient_Name Doctor_ID Slot
* DOCTOR Doctor_ID
* PATIENT Patient_Name
* SUMMARY

Sample Input
* 3
* D101 Anitha Cardiology 3
* D102 Rahul Neurology 2
* D103 Meera Dermatology 4
* 7
* BOOK Arun D101 10AM
* BOOK Priya D101 11AM
* BOOK Arun D102 2PM
* BOOK Arun D101 10AM
* CANCEL Arun D101 10AM
* PATIENT Arun
* SUMMARY

Sample Output
* Appointment Booked
* Appointment Booked
* Appointment Booked
* Appointment Already Exists
* Appointment Cancelled
* Appointments for Arun - D102 - 2PM
* Total Appointments: 2
* Most Booked Doctor: D101
* Available Slots: 7

Additional Requirements

The program should correctly handle:
* Booking a slot that does not exist.
* Booking an already occupied slot.
* Booking the same doctor's same slot by another patient.
* Cancelling an appointment.
* Cancelling a non-existing appointment.
* Multiple appointments for the same patient.
* Multiple doctors having the same number of appointments.
* Doctors with zero available slots.
* Multiple operations performed sequentially.

The program must use classes and objects for implementing the system.




```python
# APPOINTMENT MANAGEMENT SYSTEM

class Doctor:
  def __init__(self,doctor_id,doctor_name,specialization,available_slots):
      self.doctor_id = doctor_id
      self.doctor_name = doctor_name
      self.specialization = specialization
      self.available_slots = available_slots


class AppointmentSystem:
  def __init__(self):
      self.doctors = {}
      self.appointments = []

# Add doctor to the system
  def add_doctor(self, doctor):
      self.doctors[doctor.doctor_id] = doctor


# Book Appointment.
  def book_appointment(self,patient,doctor_id,slot):

# Check whether the doctor exists.
    if doctor_id not in self.doctors:
      print("Doctor not found.")
      return
# Check whether the patient already booked the appointment with the same doctor.
    for appointment in self.appointments:
      if (appointment["patient"] == patient and appointment["doctor"] == doctor_id):
        print("Appointment already exists.")
        return
# Check the slot is already occupied.
    for appointment in self.appointments:
      if (appointment["doctor"] == doctor_id and appointment["slot"] == slot):
        print("Slot not available.")
        return
# Book the appointment.
    appointment = {"patient":patient,"doctor":doctor_id,"slot":slot}
    self.appointments.append(appointment)
    print("Appointment Booked.")


# Cancel Appointment.
  def cancel_appointment(self,patient,doctor_id,slot):
    for appointment in self.appointments:
      if (appointment["patient"] == patient and
          appointment["doctor"] == doctor_id and
          appointment["slot"] == slot):
        self.appointments.remove(appointment)
        print("Appointment Cancelled")
        return
        print("Appointment Not Found")

# Display doctor details
  def display_doctor(self, doctor_id):

    if doctor_id not in self.doctors:
      print("Doctor Not Found")
      return

    doctor = self.doctors[doctor_id]

    # Count current appointments
    count = 0

    for appointment in self.appointments:
      if appointment["doctor"] == doctor_id:
          count += 1

    available = doctor.available_slots - count

    # Check Doctor's capacity
    doctor = self.doctors[doctor_id]

    if count >= doctor.available_slots:
      print("Slot not available.")
      return

    print("Doctor ID:", doctor.doctor_id)
    print("Doctor Name:", doctor.doctor_name)
    print("Specialization:", doctor.specialization)
    print("Number of available slots:", available)

# Display Patient Appointments.
  def display_patient(self,patient):
    print("Patient Appointments:")
    for appointment in self.appointments:
      if appointment["patient"] == patient:
        doctor = self.doctors[appointment["doctor"]]
        print("Doctor Name:", doctor.doctor_name)
        print("Specialization:", doctor.specialization)
        print("Slot:", appointment["slot"])
        print()
      else:
        print("No Appointments Found")
        return
# Appointment Summary.
  def summary(self):
    total = len(self.appointments)
    print("Total Appointments:", total)

# Count Appointments for each doctor.
    doctor_counts = {}
    for appointment in self.appointments:
      doctor_id = appointment["doctor"]
      if doctor_id not in doctor_counts:
        doctor_counts[doctor_id] = 0
      doctor_counts[doctor_id] += 1

# Doctor with highest number of appointments.
    if len(doctor_counts) == 0:
      print("Most booked Doctor : None")
    else:
      most_booked_doctor = None
      highest_count = 0

    for doctor_id in self.doctors:
      count = doctor_counts.get(doctor_id, 0)

      if count > highest_count:
        highest_count = count
        most_booked_doctor = doctor_id

    print("Most Booked Doctor:", most_booked_doctor)

# Calculate total available slots across all doctors.
    total_available = 0

    for doctor_id in self.doctors:
      doctor = self.doctors[doctor_id]
      count = 0

      for appointment in self.appointments:
          if appointment["doctor"] == doctor_id:
              count += 1
      available = doctor.available_slots - count
      total_available += available
    print("Available Slots:", total_available)

# ----------------------------MAIN------------------------------#

# Number of doctors
n = int(input("Number of Doctors : "))

# Create appointment system
system = AppointmentSystem()

# Read doctor details
for i in range(n):

    data = input().split()  # Split the data that is read from the user and add it to specific indexes.

    doctor_id = data[0]
    doctor_name = data[1]
    specialization = data[2]
    available_slots = int(data[3])

    doctor = Doctor(doctor_id,doctor_name,
                    specialization,available_slots)
    system.add_doctor(doctor)


# Number of operations
m = int(input("Enter the number of operations to perform : "))


# Process operations
for i in range(m):
    data = input().split()
    operation = data[0]
    if operation == "BOOK":
        patient = data[1]
        doctor_id = data[2]
        slot = data[3]
        system.book_appointment(patient,doctor_id,slot)

    elif operation == "CANCEL":
        patient = data[1]
        doctor_id = data[2]
        slot = data[3]
        system.cancel_appointment(patient,doctor_id,slot)

    elif operation == "DOCTOR":
        doctor_id = data[1]
        system.display_doctor(doctor_id)

    elif operation == "PATIENT":
        patient = data[1]
        system.display_patient(patient)

    elif operation == "SUMMARY":
        system.summary()







```
