```python
# To create an empty array of size 4x4 with int values.
# np.empty((rows,columns),datatype)

empty_array = np.empty((4,3))  # Return a 4x3 matrix with random values of float datatype.
print("Array")
print(empty_array)
new_empty_array = np.empty((4,3),int)  # Return a 4x3 matrix with random values of int datatype.
print("New Array")
print(new_empty_array)

```

```python
# ones and zeros

sample = np.ones((3,4),int)
print("1-matrix")
print(sample)

sample1 = np.zeros((3,4),int)
print("0-matrix")
print(sample1)

sample2 = np.eye(4,4)
print("matrix")
print(sample2)
```

```python
a = np.arange(1,21)
print(a)
new_a = a.reshape(4,5)
print(new_a)
print(new_a.flatten)  # Convert the matrix - to single dimensionality.
```

```python
mat1 = np.array([[44,55],[66,77]])
mat2 = np.array([[10,66],[5,90]])
print("\n ADDITION")
Addition = np.add(mat1,mat2)
print(Addition)

print("\n SUBTRACION")
Subtraction = np.subtract(mat1,mat2)
print(Subtraction)
print(f"\n Absolute value")
print(np.abs(Subtraction))

print("\n MULTIPLICATION")
print(np.multiply(mat1,mat2))

print("\n DIVISION")
print(np.divide(mat1,mat2))


print("\n Sum of row wise elements")
print(np.sum(mat1, axis=1))  # Row Wise Sum - axis = 1.

print("\n Sum of Column wise elements")
print(np.sum(mat1, axis=0))  # Column Wise Sum - axis = 0.
```
