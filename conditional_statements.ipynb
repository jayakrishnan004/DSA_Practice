Write a program check whether the number is even or odd.

```python
a = float(input("Enter the number:"))
if a%2 == 0:
  print("The number is even..")
else:
  print("The number is odd..")
```

Read a numerical value from the user and check whether the number is completely divisible by 7 or not.


```python
num = float(input("Enter the enumber : "))
if num%7 == 0:
  print(f"{num} is completely divisible by 7.")
else:
  print(f"{num} is not completely divisible by 7.")
```

Accept 2 numbers from the user compare them and print which number is larger. If they are equal, print both numbers are equal.

```python
num1 = int(input("Enter the first number:"))
num2 = int(input("Enter the second number:"))
if num1>num2:
  print(f"{num1} is larger..")
elif num1<num2:
  print(f"{num2} is larger..")
else:
  print(f"Both numbers are equal.")
```

Read a student score as an input, the score can be any value from 0 to 100. Assign grades based on the following scale.
90 and above A grade, (80-89) B grade, (70 - 79) C grade, (60-69) D grade, below 60 F grade.

```python
name = input("Enter the name:")
score = int(input("Enter the score:"))
if score>= 90:
  print(f"Congrats {name}, you scored {score} marks and got A Grade.")
elif score >= 80:
  print(f"Congrats {name}, you scored {score} marks and got B Grade.")
elif score>70:
  print(f"Congrats {name}, you scored {score} marks and got C Grade.")
elif(score>60):
  print(f"{name} scored {score} marks and got D Grade.")
else:
  print(f"{name} scored {score} marks and got F Grade and you failed in the exam.")
```

```python
month = input("Enter the month:")
match month:
  case "January":
    print("31 days.")
  case "February":
    print("28 days")
  case "March":
    print("31 days")
  case "April":
    print("30 days")
  case "May":
    print("31 days")
  case _: #to print default case
    print("Invalid entry")
```
