# **EXCEPTION HANDLING**

```python
# ValueError

try:
    age = int(input("Enter the age : "))
    print(f"You will be {age+1} year next year.")
except ValueError:
    print(f"Oops.. Enter a valid number.")
```

```python
# ZeroDivisionError

try:
    num1 = int(input("Enter first number : "))
    num2 = int(input("Enter the second number :"))
    print(f"Division : {num1/num2}")
except ZeroDivisionError:
    print("Cannot divisible by zero...!")
```

#**OOPS**
* in oops data is stored in fields called as ATTRIBUTES.
* class - blueprint for creating objects.
* object - it is the instance of a class. Objects are created for accessing class.
* Methods - procedures to do ->functions.


```python
# CLASS - Structure

# class classname:
#     constructor:
#         attributes[variables to store data]
#     def methods():
```

```python
# EXAMPLE

class Book:
    def __init__(self, num_of_pages, num_of_chapters, title, author, price):  # Constructor.
        self.num_of_pages = num_of_pages
        self.num_of_chapters = num_of_chapters
        self.title = title
        self.author = author
        self.price = price

    def display_details(self):
        print(f"The book name is {self.title}, the author is {self.author}. Price of the book is {self.price}.")

book1 = Book(num_of_pages=120, num_of_chapters=5, title="jk123", author="jk", price=360 )  # Object.
book1.display_details()  # Object calling.

book2 = Book(num_of_pages=200, num_of_chapters=12, title="THE ALCHEMIST", author="Paulo Coelho", price=350)
book2.display_details()
```

```python
#  EXAMPLE 2

class Car:
    def __init__(self, colour, brand, year, model):
        self.colour = colour
        self.brand = brand
        self.year = year
        self.model = model
    def display_details(self):
        print(f"This car is {self.brand} {self.model}. The colour of the car is {self.colour} and the year of manufacturing is {self.year}.")

car1 = Car("Black", "Toyota", 2022, "Innova")
car1.display_details()

car2 = Car("White", "Maruti", 2022, "Baleno")
car2.display_details()
```

```python
# EXAMPLE 3
# Area of a Rectangle

class Rectangle:
    def __init__(self, length,breadth):
        self.length = length
        self.breadth = breadth
    def area(self):
        area = self.length*self.breadth
        print(f"The area of the Rectangle is {area}")
rec1 = Rectangle(5,8)
rec1.area()
```

**Class Attributes**
* Attributes can belong to individual instances(Instance Attributes) or be shared across all the instances of the class.

```python
class student:
    school_name = "SJCET"  # Class Attribute

    def __init__(self, name, grade):  # Instance Attribute.
        self.name = name
        self.grade = grade
    # To change the class attribute, use decorative - @classmethod

    @classmethod
    def change_class(cls, new_name):
        cls.school_name = new_name

print(f"Old Name : {student.school_name}")  # Class_name.class_attribute

student.change_class("SJCET PALAI")
print(f"Changed Name : {student.school_name}")
```

# **INHERITNCE**

* used to reuse from another class
* code reusability
* allows a new class to inherit attributes and methods from an existing class(parent/base class).

```python
# Example

class Animal :
    def __init__(self, name):
        self.name = name
class Dog(Animal):
    def sound(self):
        print(f"The dog {self.name} barks.")
class Cat(Animal):
    def sound(self):
        print(f"The cat {self.name} meows.")
dog1 = Dog("Tippu")
cat1 = Cat("Poocha")

dog1.sound()
cat1.sound()
```
