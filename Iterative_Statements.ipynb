Find the product of n consecutive natural numbers.

```python
n = int(input("Enter the value of n : "))
product = 1
for i in range(1,n+1):
  product = product*i
print(product)
```

given an integer num = 4523, write a while loop, calculate the sum of its individual numbers.

```python
num = 4523
sum = 0
while(num>0):
  sum += num%10 #or sum = sum + num%10  #takes the remainder
  num = num//10 #removes the last digit - 452..45..4..0, when it reaches 0, while loop closes.
print(sum)
```
