# **PANDAS**
Pandas is a fast and easy to use powerful opensource library built on top of the python programming language. It is primarily used for data analysis, data manipulation and data cleaning.

* In pandas, Series and DataFrame are the two fundamental data structures used for data manipulation and analysis. We can think a SERIES representing a single column while a DATAFRAME representing an entire multi column table.

```python
import pandas as pd
print(pd.__version__)
```

```python
# Series from pandas.
# Series - one dimensional labelled array.

list =[34,45,27]
series = pd.Series(list)
new_series = pd.Series(list,index=['11AM','1PM','7PM'])
print("SERIES")
print(series)
print("\nNEW SERIES")
print(new_series)



s1 = pd.Series([])  # Create an empty series.
print("\nEMPTY SERIES")
print(s1)


#  Creating a series from a dictionary.

D1 = pd.Series({"p":1, "q":2})
print("\nSERIES FROM DICTIONARY")
print(D1)
```

```python
# Read CSV File
filepath = "/content/drive/MyDrive/DSA - ICTAK/DATASETS/retail_sales_pandas_training_dataset (1).csv"
df = pd.read_csv(filepath)
print(df)

# To understand the loaded data.
print("\nCOLUMNS")
print(df.columns)  # Return the column names of the data.

print("\nSHAPE")
print(df.shape)  # Return the shape of the data.

print("\nSIZE")
print(df.size)  # Return the size of the data.

print("\nTop 5\n")
print(df.head())  # Return the top 5 data.

print("\nTop 3\n")
print(df.head(3))  # specify the number if we want the data specifically..

print("\nLAST 5\n")
print(df.tail())  # Return the last 5 data.

print("\nLAST 10\n")
print(df.head(10))  # specify the number if we want the data specifically..

print("\nDESCRIBE\n")
print(df.describe())  # Will give the overall understanding of the data.

print("\nDATA TYPES\n")  # Give the data types of each columns.
print(df.dtypes)

print("\nIS NULL\n")  # Check whether there is a null value.
print(df.isnull())

df['D.o.J'] = '13-Feb-2026'  # To add a new column with a specific value for all records.
print("\nADD A NEW COLUMN\n")
print(df)

df.drop('D.o.J', axis=1, inplace = True)
print("\nREMOVE A COLUMN\n")
print(df)

```

# **AGGREGATE FUNCTIONS**

```python
print("\nMean\n")
print(df["Revenue"].mean())

print("\nSUM\n")
print(df["Revenue"].sum())

print("\nMAXIMUM\n")
print(df["Revenue"].max())

print(f"COUNT : {df["Revenue"].count()}")
```

# **SORTING AND FILTERING**

```python
# To filter out the sales happening in east region.
print(df[df["Region"]=="East"])
```

```python
# To filter out the sales for laptop.
print(df[df["Product"]=="Laptop"])
```

```python
# To filter out the sales done by Eva for a revenue greater than 10000.
new_data = df[(df["Sales_Rep"]== "Eva") & (df["Revenue"]> 10000)]
print(new_data)
```

```python
# To filter out the sales done by Alice for the product "Laptop"

Lap = df[(df["Sales_Rep"]== "Alice") & (df["Product"] == "Laptop")]
print(Lap)
```
