Initialize a class person with the constructor to initialize the name and the adhaar number, children and age should set as 0. We have to define get_name() to print the name field, define get_adhaar() to print adhaar number, define set_age to assign a new value to age field also define senior_citizen() which checks for the age>60. Finally create an object for the class and call the respective methods.

```python
class person:
  def __init__(self):
   self.name = 0
   self.adhaar = 0
   self.children = 0
   self.age = 0
  def get_name(self):
    self.name = input("Enter the name of the person :")
    print(f"The name of the person is {self.name}")
  def get_adhaar(self):
    self.adhaar = int(input("Enter the adhaar number : "))
    print(f"The Adhaar number of the person is {self.adhaar}")
  def set_age(self):
    self.age = int(input("Enter the age :"))
    print(f"Age = {self.age}")
  def senior_citizen(self):
    if self.age>60:
      print("Senior Citizen")
    else:
      print("Not a senior citizen")

p1 = person()
p1.get_name()
p1.get_adhaar()
p1.set_age()
p1.senior_citizen()
```

Create a class circle with a constructor, initializing the radus r, another function to find the area of the circle, another function to find circumference of the circle.

```python
class circle:
  def __init__(self, rad):
    self.rad = rad
  def area(self):
    self.area = 3.14*self.rad*self.rad
    print(f"The area of the circle is {self.area} sq. unit")
  def cir(self):
    self.cir = 2*3.14*self.rad
    print(f"The circumference of the circle is {self.cir} sq. unit")

rad = int(input("Enter the radius of the circle : "))

c = circle(rad)
c.area()
c.cir()
```

There are 2 classes namely num1 and num2. In num1 class, there are 2 variables A and B that needs to be initialize using constructor. It also have a member function show_value to display this value. num2 class is inherited from num1 and it has 2 functions

*   show_sum() display the sum of the variables
*   show_product() to display the product of the variables



```python
class num1:
  def __init__(self, a, b):
    self.a = a
    self.b =b
  def show_value(self):
    print(f"The values of the variables are a = {self.a} and b = {self.b}")
class num2(num1):
  def __init__(self, a, b):
    super().__init__(a,b)
  def show_sum(self):
    self.show_sum = self.a + self.b
    print(f"The sum of the numbers is {self.show_sum}")
  def show_product(self):
     self.show_product = self.a * self.b
     print(f"The product of the numbers is {self.show_product}")
a = int(input("Enter the num a :"))
b = int(input("Enter the num b :"))

c = num2(a,b)
c.show_value()
c.show_sum()
c.show_product()

```

There are 2 classes student and the master_student. Student is the base class and master_student is the derived class. The class student has 2 member variables id and mark. There is a constructor to initialize these variables and a function show_result() which will check the mark for passed and failed. if the mark >=40 it is Pass otherwise it is Fail. Similarly in the class master_student contain a function show_result() that will check whether the mark is >= 50. If that is the case then it is a Pass otherwise it will be a Fail. Initialize the student object and a master_student object and call the functions respectively.



```python
class student:
  def __init__(self, id, mark):
    self.id = id
    self.mark = mark
  def show_result(self):
    if self.mark >= 40:
      print(f"Pass")
    else:
      print("Fail")
class master_student(student):
  def __init__(self, id, mark):
    super().__init__(id, mark)
  def show_result(self):
    super().show_result()
    if self.mark >= 50:
        print(f"Pass")
    else:
        print("Fail")


id = int(input("Enter the student_id :"))
mark = int(input("Enter the mark of the student : "))

s1 = master_student(id, mark)
s1.show_result()
```

**POLYMORPHISM EXAMPLE**

```python
# Polymorphism

class Polygon:
  def __init__(self):
    pass
  def area(self):
    pass
  def perimeter(self):
    pass

class Rectangle(Polygon):
  def __init__(self,l,b):
    self.len = l
    self.bred = b
  def area(self):
    return(self.len*self.bred)
  def perimeter(self):
    return(2*(self.len*self.bred))
class Triangle(Polygon):
  def __init__(self, p, q, r):
    self.a = p
    self.b = q
    self.c = r
  def area(self):
    s = (self.a+self.b+self.c)/2
    return ((s*(s-self.a)*(s-self.b)*(s-self.c))**0.5)
  def perimeter(self):
    self.a+self.b+self.c

r1 = Rectangle(5,8)
t1 = Triangle(3,4,5)
print(r1.area())
print(r1.perimeter())
print(t1.area())
print(t1.perimeter())
```
