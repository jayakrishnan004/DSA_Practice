```python
class employee:
    def __init__(self):
        self.name="Rahul"  # Public
        self._salary=5000  #protected
        self.__password="12345"  #private
    def show_details(self):
        print(self.name)
        print(self._salary)
        print(self.__password)

emp =employee()
print(emp.name)
print(emp._salary) # in the same class we can print the private things
emp.show_details()
```

```python
class Employee:
  def __init__(self):
    self.name = "Anu"  # Public
    self._salary = 50000  # Protected
    self.__password = "12345"  # Private

  def show_parent(self):
    print("Name : ", self.name)
    print("salary : ", self._salary)
    print("Password : ", self.__password)



class manager(Employee):
  def show_manager(self):
    print("Name : ", self.name)
    print("Salary : ", self._salary)
    # print("Password : ", self.__password)
    # Cannot print the private things in another class, in the same class it can print.

man = manager()
man.show_manager()
```
