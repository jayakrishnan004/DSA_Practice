DICTIONARY
key- value pair - its mutable in nature

```python
student = {
    "name": "JK",
    "age" : 22,
    "place" : "kottayam"
}
print(student)
print(student["place"]) #to fetch or read value using the key
print(student.get("place")) #to fetch the value using get()

print(student.get("course")) #return none if the mentioned key is not present

student["name"] = "jeswin"
print(student)
student["course"] = "BA. Literature" # To add a new key value pair to the dictionary
print(student)
del student["age"] # to delete from the dictionary
print(student)
del student["course"]
print(student)


#to add age and course using update
student.update({"age":22, "course": "data science and analytics"})
print(student)


print(student.pop("course"))
print(student.pop("course","english")) #if there is no such element in the code, default value prints - ie. English

print(student.keys())
print(student.values())

student["name"] = "jayakrishnan"
print(student)

for key in student:  #to list the key from the dictionary , if the "key" is mentioned as k in the case, it dont works.
  print(key)

for value in student.values(): #to list the values from the dictionary
  print(value)

#to list the keys and values in the pattern key:value
for key,value in student.items():
  print(f"{key}:{value}")
```

LIST is  a built in mutable, and ordered collection of items enclosed in [].

```python
a = [1,2,3,4,5]
print(a)
b = ["apple", "banana"]
print(b)

c = list((1,2,3,4, "apple")) #use2 brackets, ow it leads to error
print(c)

#creating list with repeated elements
d = [5]*4
print(d)

e = [1,6,3,4,5]
print(e[1])
print(e[-1])


#Adding elements
a = [1,2,3]
a.append(4) #append element at last
print(a)

a.insert(2,6) #insert element at specific position
print(a)

a.extend([7,8]) #add multiple elements at the end of the list
print(a)

#updating elements
a[2] = 10
print(a)

#remove the first occurence of an element
a.remove(10)
print(a)

#deletes element at specific position or the last element if no specific index is given
a.pop()
print(a)
a.pop()
print(a)
a.pop()
print(a)

#delete element at specific position
del a[0]
print(a)

#clear all items
a.clear()
print(a)


z = ["apple", "banana", "blueberry"]
for item in z:
  print(item)

z[1:2] = ["orange", "kiwi"]
print(z)

#SORTING
r = [11,55,44,63,88,2,9,7]
r.sort() #Ascending
print(r)

r.sort(reverse = True) #Descending
print(r)

```

To display the years of experience of the employee with code 102. Given the following data.


*   Employee1 - empcode - 101 , name - Reshma , Years_exp - 8 , dept - HR
*   Employee2 - empcode - 102 , name - John , Years_exp - 7 , dept - Administraion
* Employee3 - empcode - 103 , name - Agarwal , Years_exp - 11 , dept - Research







```python
employee = [
    {"empcode": 101, "name": "Reshma", "experience": 8, "dept": "HR"},
    {"empcode": 102, "name": "John", "experience": 7, "dept": "Administration"},
    {"empcode": 103, "name": "Agarwal", "experience": 11, "dept": "Research"}
]

for i in employee:
  if i["empcode"] == 102:
    print(f"The employee is {i["name"]} has {i["experience"]} years of experience.")
```

**SET**  - non primitive, unordered and it won't allow duplicate values.

```python
marks = {23, 29, 37, 19}
print(marks)
print(type(marks))

marks.add(55)  # Add one new element in the set
print(marks)

for mark in marks:
  print(mark)

marks.update({200, 633})  # If we want to add more elements
print(marks)

marks.remove(633)  # Remove one element from the set
print(marks)


marks.discard(23)
print(marks)


marks.pop()  # Cannot remove a specific element, pop() will remove the elements arbitary
print(marks)
```
