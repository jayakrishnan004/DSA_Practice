To display the years of experience of the employee with code 102. Given the following data.

Employee1 - empcode - 101 , name - Reshma , Years_exp - 8 , dept - HR
Employee2 - empcode - 102 , name - John , Years_exp - 7 , dept - Administraion
Employee3 - empcode - 103 , name - Agarwal , Years_exp - 11 , dept - Research

```python
employee = [
    {"empcode":101, "name":"reshma", "yrs":8, "dept":"HR"},
    {"empcode":102, "name":"john", "yrs":7, "dept":"adm"},
    {"empcode":103, "name":"Agarwal", "yrs":11, "dept":"Research"}
]

for i in employee:
  if i["empcode"] == 102:
    print(f"{i["name"]} has {i["yrs"]} years of experience")
```

```python
name = input("Enter the name:")
score = int(input("Enter the score:"))
if score>= 90:
  print(f"Congrats {name}, you scored {score} marks and got A Grade.")
elif score >= 80:
  print(f"Congrats {name}, you scored {score} marks and got B Grade.")
elif score>70:
  print(f"Congrats {name}, you scored {score} marks and got C Grade.")
elif(score>60):
  print(f"{name} scored {score} marks and got D Grade.")
else:
  print(f"{name} scored {score} marks and got F Grade and you failed in the exam.")
```

```python
month = input("Enter the month:")
match month:
  case "January":
    print("31 days.")
  case "February":
    print("28 days")
  case "March":
    print("31 days")
  case "April":
    print("30 days")
  case "May":
    print("31 days")
  case _: #to print default case
    print("Invalid entry")
```

```python
# write a program that prints all even numbers from 2 upto 20 using a while loop.

i = 2
while(i<=20):
  if i% 2 == 0:
    print(i)
  i = i+1
```

```python
# Given an integer num = 4523, write a while loop, calculate the sum of its individual numbers.

num = 4523
sum = 0
while(num > 0):
  sum = sum+num%10
  num = num//10
print(sum)
```

```python
class Student:  # Class is the blueprint from which object is created
    def __init__(self, name, age, course):
        self.name = name
        self.age = age
        self.course = course

    def display_details(self):  # Method
        print("Name:", self.name)
        print("Age:", self.age)
        print("Course:", self.course)


student1 = Student("John", 21, "AI and Data Science")  # Object is the instance of a class

student1.display_details()  # Object calling
```
